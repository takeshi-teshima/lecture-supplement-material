```python
import random
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
import pytorch_lightning as pl
from pytorch_lightning.callbacks import EarlyStopping, Callback
from torch.utils.data import DataLoader, TensorDataset
from sklearn.datasets import make_blobs
from sklearn.model_selection import train_test_split
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap

# -----------------------------------------
# Fix Random Seeds
# -----------------------------------------
SEED = 94
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

# -----------------------------------------
# Original Color Palette
# -----------------------------------------
color_palette = {
    "white": "#FFFFFF",
    "light_gray": "#D3D3D3",
    "gray": "#808080",
    "black": "#000000",
    "green": "#008000",
    "blue": "#0000FF",
    "light_blue": "#ADD8E6",
    "light_light_blue": "#E0FFFF",
    "yellow": "#FFFF00",
    "orange": "#FFA500",
    "dark_orange": "#FF8C00",
    "purple": "#800080",
}

# -----------------------------------------
# Class Colors (5 Classes)
# -----------------------------------------
# We will use these both for data points AND region coloring
class_colors = [
    color_palette["dark_orange"],  # Class 0
    color_palette["blue"],         # Class 1
    color_palette["green"],        # Class 2
    color_palette["purple"],       # Class 3
    color_palette["yellow"]        # Class 4
]

# -----------------------------------------
# Create a ListedColormap from class_colors
# so that region i matches class i
# -----------------------------------------
region_cmap = ListedColormap(class_colors)

# -----------------------------------------
# Generate Synthetic Data (2D, 5 classes)
# -----------------------------------------
X, y = make_blobs(
    n_samples=1000,
    centers=5,
    n_features=2,
    random_state=SEED,
    cluster_std=1.5
)

# Split into Train (70%), Val (15%), Test (15%)
X_temp, X_test, y_temp, y_test = train_test_split(X, y, test_size=0.15, random_state=SEED)
X_train, X_val, y_train, y_val = train_test_split(X_temp, y_temp, test_size=0.1765, random_state=SEED)
# 0.1765 ~ 0.15 / 0.85 => overall 70/15/15 split

# Convert to tensors
X_train_t = torch.tensor(X_train, dtype=torch.float32)
y_train_t = torch.tensor(y_train, dtype=torch.long)
X_val_t   = torch.tensor(X_val,   dtype=torch.float32)
y_val_t   = torch.tensor(y_val,   dtype=torch.long)
X_test_t  = torch.tensor(X_test,  dtype=torch.float32)
y_test_t  = torch.tensor(y_test,  dtype=torch.long)

# -----------------------------------------
# LightningDataModule for the 3 splits
# -----------------------------------------
class MyDataModule(pl.LightningDataModule):
    def __init__(self, X_train, y_train, X_val, y_val, X_test, y_test, batch_size=32):
        super().__init__()
        self.X_train = X_train
        self.y_train = y_train
        self.X_val   = X_val
        self.y_val   = y_val
        self.X_test  = X_test
        self.y_test  = y_test
        self.batch_size = batch_size

    def setup(self, stage=None):
        self.train_dataset = TensorDataset(self.X_train, self.y_train)
        self.val_dataset   = TensorDataset(self.X_val,   self.y_val)
        self.test_dataset  = TensorDataset(self.X_test,  self.y_test)

    def train_dataloader(self):
        return DataLoader(self.train_dataset, batch_size=self.batch_size, shuffle=True)

    def val_dataloader(self):
        return DataLoader(self.val_dataset, batch_size=self.batch_size, shuffle=False)

    def test_dataloader(self):
        return DataLoader(self.test_dataset, batch_size=self.batch_size, shuffle=False)

data_module = MyDataModule(X_train_t, y_train_t, X_val_t, y_val_t, X_test_t, y_test_t, batch_size=32)

# -----------------------------------------
# 4-layer Feedforward Model
# -----------------------------------------
class FeedForwardNet(pl.LightningModule):
    """
    A simple 4-layer feedforward network:
      - Input layer: 2 features
      - Hidden layer 1: 64 units, ReLU
      - Hidden layer 2: 32 units, ReLU
      - Output layer: 5 units (5 classes)
    Uses CrossEntropyLoss, Adam, logs train/val metrics.
    We'll store test metrics in separate lists each epoch.
    """
    def __init__(self, input_dim=2, hidden1=64, hidden2=32, output_dim=5, lr=1e-3):
        super().__init__()
        self.save_hyperparameters()

        self.net = nn.Sequential(
            nn.Linear(input_dim, hidden1),
            nn.ReLU(),
            nn.Linear(hidden1, hidden2),
            nn.ReLU(),
            nn.Linear(hidden2, output_dim)
        )

        # For storing batch-level metrics
        self.training_losses = []
        self.validation_losses = []
        self.training_accs = []
        self.validation_accs = []

        # For storing final per-epoch test metrics
        self.test_loss_per_epoch = []
        self.test_acc_per_epoch = []

    def forward(self, x):
        return self.net(x)

    def training_step(self, batch, batch_idx):
        x, y = batch
        logits = self(x)
        loss = nn.CrossEntropyLoss()(logits, y)
        preds = torch.argmax(logits, dim=1)
        acc = (preds == y).float().mean()

        self.training_losses.append(loss.item())
        self.training_accs.append(acc.item())
        return loss

    def validation_step(self, batch, batch_idx):
        x, y = batch
        logits = self(x)
        loss = nn.CrossEntropyLoss()(logits, y)
        preds = torch.argmax(logits, dim=1)
        acc = (preds == y).float().mean()

        self.validation_losses.append(loss.item())
        self.validation_accs.append(acc.item())

        self.log("val_loss", loss, prog_bar=True)
        return loss

    def configure_optimizers(self):
        return optim.Adam(self.parameters(), lr=self.hparams.lr)

# -----------------------------------------
# Manual Test Pass
# -----------------------------------------
def compute_test_metrics(pl_module, trainer):
    """
    Computes average test loss & accuracy by iterating over test dataloader.
    """
    test_loader = trainer.datamodule.test_dataloader()
    pl_module.eval()
    losses, accs = [], []
    with torch.no_grad():
        for batch in test_loader:
            x, y = batch
            logits = pl_module(x)
            loss = nn.CrossEntropyLoss()(logits, y)
            preds = torch.argmax(logits, dim=1)
            acc = (preds == y).float().mean()

            losses.append(loss.item())
            accs.append(acc.item())
    pl_module.train()
    return np.mean(losses), np.mean(accs)

# -----------------------------------------
# Callback: Evaluate on Test after each epoch
# -----------------------------------------
class EvaluateOnTestCallback(Callback):
    def on_validation_end(self, trainer, pl_module):
        test_loss, test_acc = compute_test_metrics(pl_module, trainer)
        pl_module.test_loss_per_epoch.append(test_loss)
        pl_module.test_acc_per_epoch.append(test_acc)

# -----------------------------------------
# Plot Helper Functions
# -----------------------------------------
def plot_data_scatter(X_data, y_data, title="Data Scatter Plot"):
    """
    Scatter of 2D data with class-based coloring.
    """
    plt.figure(figsize=(6,5))
    for c in range(5):
        idxs = (y_data == c)
        plt.scatter(X_data[idxs, 0], X_data[idxs, 1],
                    color=class_colors[c],
                    edgecolors=color_palette["black"],
                    label=f"Class {c}")
    plt.title(title)
    plt.xlabel("x1")
    plt.ylabel("x2")
    plt.legend()
    plt.show()

def plot_decision_regions_matched_colors(model, X_data=None, y_data=None,
                                         title="Decision Boundary", show_data=True):
    """
    Plots the decision boundaries with the same colors as the data points.
    We auto-compute the mesh range to include all X_data (with padding).
    """
    if X_data is not None and len(X_data) > 0:
        x_min = X_data[:,0].min() - 2.0
        x_max = X_data[:,0].max() + 2.0
        y_min = X_data[:,1].min() - 2.0
        y_max = X_data[:,1].max() + 2.0
    else:
        # Fallback if no data
        x_min, x_max = -12, 12
        y_min, y_max = -12, 12

    step = 0.1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, step),
                         np.arange(y_min, y_max, step))
    grid_points = np.c_[xx.ravel(), yy.ravel()]
    grid_torch = torch.tensor(grid_points, dtype=torch.float32)
    
    with torch.no_grad():
        logits = model(grid_torch)
        preds = torch.argmax(logits, dim=1)
    Z = preds.numpy().reshape(xx.shape)

    # We define discrete levels for contourf
    levels = np.arange(-0.5, 5, 1)

    plt.figure(figsize=(6,5))
    # Fill the regions with region_cmap
    # alpha=0.3 to get a lighter shade
    plt.contourf(xx, yy, Z, levels=levels, cmap=region_cmap, alpha=0.3)

    if show_data and X_data is not None and y_data is not None:
        for c in range(5):
            idxs = (y_data == c)
            plt.scatter(X_data[idxs, 0], X_data[idxs, 1],
                        color=class_colors[c],
                        edgecolors=color_palette["black"])

    plt.title(title)
    plt.xlabel("x1")
    plt.ylabel("x2")
    plt.show()

# -----------------------------------------
# Instantiate Model and Callbacks
# -----------------------------------------
model = FeedForwardNet(input_dim=2, hidden1=64, hidden2=32, output_dim=5, lr=1e-3)
early_stop_callback = EarlyStopping(monitor="val_loss", patience=10, mode="min", verbose=True)
test_callback = EvaluateOnTestCallback()

# -----------------------------------------
# BEFORE Training
# -----------------------------------------
plot_data_scatter(X_train, y_train, title="Train Data Scatter Plot")

plot_decision_regions_matched_colors(
    model, X_data=X_train, y_data=y_train,
    title="Decision Boundary BEFORE Training (with Train Data)",
    show_data=True
)

plot_decision_regions_matched_colors(
    model,
    title="Decision Boundary BEFORE Training (no Train Data)",
    show_data=False
)

# -----------------------------------------
# Train the Model
# -----------------------------------------
trainer = pl.Trainer(
    max_epochs=100,
    callbacks=[early_stop_callback, test_callback],
    deterministic=True,
    logger=False,              # no default logger
    enable_checkpointing=False
)
trainer.fit(model, datamodule=data_module)

# -----------------------------------------
# Collect Metrics
# -----------------------------------------
train_batches_per_epoch = len(data_module.train_dataloader())
val_batches_per_epoch   = len(data_module.val_dataloader())

def batch_to_epoch(values_list, num_batches_per_epoch):
    epoch_values = []
    idx = 0
    while idx < len(values_list):
        batch_slice = values_list[idx : idx + num_batches_per_epoch]
        epoch_values.append(np.mean(batch_slice))
        idx += num_batches_per_epoch
    return epoch_values

train_loss_per_epoch = batch_to_epoch(model.training_losses, train_batches_per_epoch)
val_loss_per_epoch   = batch_to_epoch(model.validation_losses, val_batches_per_epoch)
test_loss_per_epoch  = model.test_loss_per_epoch

train_acc_per_epoch = batch_to_epoch(model.training_accs, train_batches_per_epoch)
val_acc_per_epoch   = batch_to_epoch(model.validation_accs, val_batches_per_epoch)
test_acc_per_epoch  = model.test_acc_per_epoch

# Misclassification
train_misclass = [1.0 - acc for acc in train_acc_per_epoch]
val_misclass   = [1.0 - acc for acc in val_acc_per_epoch]
test_misclass  = [1.0 - acc for acc in test_acc_per_epoch]

# Align if early stopping
min_len = min(len(train_loss_per_epoch), len(val_loss_per_epoch), len(test_loss_per_epoch))
train_loss_per_epoch = train_loss_per_epoch[:min_len]
val_loss_per_epoch   = val_loss_per_epoch[:min_len]
test_loss_per_epoch  = test_loss_per_epoch[:min_len]

train_misclass = train_misclass[:min_len]
val_misclass   = val_misclass[:min_len]
test_misclass  = test_misclass[:min_len]

epochs_range = range(1, min_len + 1)

# -----------------------------------------
# Plot Loss Curves
# -----------------------------------------
plt.figure(figsize=(7,5))
plt.plot(epochs_range, train_loss_per_epoch, label="Train Loss", color=color_palette["blue"])
plt.plot(epochs_range, val_loss_per_epoch,   label="Val Loss",   color=color_palette["purple"])
plt.plot(epochs_range, test_loss_per_epoch,  label="Test Loss",  color=color_palette["dark_orange"])
plt.xlabel("Epoch")
plt.ylabel("Cross Entropy Loss")
plt.title("Loss Curves")
plt.legend()
plt.show()

# -----------------------------------------
# Plot Misclassification Curves
# -----------------------------------------
plt.figure(figsize=(7,5))
plt.plot(epochs_range, train_misclass, label="Train Misclass.", color=color_palette["blue"])
plt.plot(epochs_range, val_misclass,   label="Val Misclass.",   color=color_palette["purple"])
plt.plot(epochs_range, test_misclass,  label="Test Misclass.",  color=color_palette["dark_orange"])
plt.xlabel("Epoch")
plt.ylabel("Misclassification Rate")
plt.title("Misclassification Rate Curves")
plt.legend()
plt.show()

# -----------------------------------------
# AFTER Training: Decision Boundaries
# -----------------------------------------
plot_decision_regions_matched_colors(
    model,
    title="Decision Boundary AFTER Training",
    show_data=False
)

plot_decision_regions_matched_colors(
    model, X_data=X_train, y_data=y_train,
    title="Decision Boundary AFTER Training (with Train Data)",
    show_data=True
)

plot_decision_regions_matched_colors(
    model, X_data=X_test, y_data=y_test,
    title="Decision Boundary AFTER Training (with Test Data)",
    show_data=True
)

# --------------------------------------------------
# 追加の図：train/val/test それぞれだけの plot
# --------------------------------------------------

# Train Loss
plt.figure(figsize=(7,5))
plt.plot(epochs_range, train_loss_per_epoch, color=color_palette["blue"], label="Train Loss")
plt.title("Train Loss")
plt.xlabel("Epoch")
plt.ylabel("Cross Entropy Loss")
plt.show()

# Validation Loss
plt.figure(figsize=(7,5))
plt.plot(epochs_range, val_loss_per_epoch, color=color_palette["purple"], label="Val Loss")
plt.title("Validation Loss")
plt.xlabel("Epoch")
plt.ylabel("Cross Entropy Loss")
plt.show()

# Test Loss
plt.figure(figsize=(7,5))
plt.plot(epochs_range, test_loss_per_epoch, color=color_palette["dark_orange"], label="Test Loss")
plt.title("Test Loss")
plt.xlabel("Epoch")
plt.ylabel("Cross Entropy Loss")
plt.show()

# Train Misclassification Rate
plt.figure(figsize=(7,5))
plt.plot(epochs_range, train_misclass, color=color_palette["blue"], label="Train Misclass")
plt.title("Train Misclassification Rate")
plt.xlabel("Epoch")
plt.ylabel("Misclassification Rate")
plt.show()

# Validation Misclassification Rate
plt.figure(figsize=(7,5))
plt.plot(epochs_range, val_misclass, color=color_palette["purple"], label="Val Misclass")
plt.title("Validation Misclassification Rate")
plt.xlabel("Epoch")
plt.ylabel("Misclassification Rate")
plt.show()

# Test Misclassification Rate
plt.figure(figsize=(7,5))
plt.plot(epochs_range, test_misclass, color=color_palette["dark_orange"], label="Test Misclass")
plt.title("Test Misclassification Rate")
plt.xlabel("Epoch")
plt.ylabel("Misclassification Rate")
plt.show()
```
