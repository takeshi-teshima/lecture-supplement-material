```python
# -*- coding: utf-8 -*-

def print_utf8(text):
    print(text)
    print([f"{char}" for char in text])
    # 各文字に対してUTF-8符号化を表示
    for char in text:
        utf8_bytes = char.encode('utf-8')
        hex_repr = ' '.join(f'{byte:02x}' for byte in utf8_bytes)
        dec_repr = ' '.join(f'{byte:3d}' for byte in utf8_bytes)
        print(f"{char}: HEX: {hex_repr} | DEC: {dec_repr}")

    print([
        ''.join(f'{byte}' for byte in char.encode('utf-8'))
        for char in text
    ])
    print('')


print_utf8("いい天気だね")
```

## シェイクスピアの作品

```python
print_utf8("To be, or not to be, that is the question.")
```

## Spam

## 文字化け

```python
text = "いい天気だね"

## UTF-8で符号化
encoded_text = text.encode('utf-8')
print(encoded_text) # 16進数表記
print([''.join(f'{byte:3d}' for byte in char.encode('utf-8')) for char in text]) # 10進数表記（1文字単位で表示）

## UTF-8で符号化したものを，UTF-8で表示
print(encoded_text.decode('utf-8'))

## UTF-8で符号化したものを，SHIFT-JISで表示
print(encoded_text.decode('shift_jis', errors='replace'))
```
