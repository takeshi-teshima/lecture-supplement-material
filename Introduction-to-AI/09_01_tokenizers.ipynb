```python
import MeCab

# Initialize MeCab Tagger
tagger = MeCab.Tagger()

# Text to tokenize
text = "私はAIでデータを処理した。"

# Perform tokenization
result = tagger.parse(text)

# Display the parsed output
print("Parsed Output:")
print(result)

# Extract words as a list
words = [line.split("\t")[0] for line in result.split("\n") if line not in ("", "EOS")]
print("Tokenized Words:", "／".join(words))
```

```python
from janome.tokenizer import Tokenizer

tokenizer = Tokenizer()
text = "私はAIでデータを処理した。"
tokens = [token.surface for token in tokenizer.tokenize(text)]
print("Tokenized Words:", "／".join(tokens))
```

```python
import tiktoken

# 使用するエンコーディング（モデルに応じて選択）
encoding = tiktoken.encoding_for_model("gpt-4o")

# テキストを準備
text = "私はAIでデータを処理した。"

# テキストをトークン化
tokens = encoding.encode(text)
print("トークンID:", "／".join([str(t) for t in tokens]))

# トークンをデコード（人が読める形に戻す）
decoded_tokens = [encoding.decode([token]) for token in tokens]
print("Tokenized Words:", "／".join(decoded_tokens))
```
