```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap
from mpl_toolkits.mplot3d import Axes3D

# Generate spherical coordinates
r = np.linspace(0, 3, 50)  # Radius
theta = np.linspace(0, np.pi, 50)  # Polar angle
phi = np.linspace(0, 2 * np.pi, 50)  # Azimuthal angle
R, Theta, Phi = np.meshgrid(r, theta, phi)

# Convert spherical to Cartesian coordinates for 3D plotting
X = R * np.sin(Theta) * np.cos(Phi)
Y = R * np.sin(Theta) * np.sin(Phi)
Z = R * np.cos(Theta)

# 3D Gaussian-like density function
sigma = 1.0  # Standard deviation for Gaussian
density = np.exp(-R**2 / (2 * sigma**2))

# Custom colormap for smooth gradient (outer light blue, inner bold blue)
custom_cmap = LinearSegmentedColormap.from_list('custom_blue', ["#ADD8E6", "#0000FF"], N=256)

# Plot the 3D electron cloud orb
fig = plt.figure(figsize=(10, 10))
ax = fig.add_subplot(111, projection='3d')

# Use scatter plot with dimmed transparency
scatter = ax.scatter(X, Y, Z, c=density.ravel(), cmap=custom_cmap, alpha=density.ravel() * 0.3, s=5)

# Adjust view for a more orb-like angle
ax.view_init(elev=30, azim=60)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Single point coordinates
X, Y, Z = [0], [0], [0]  # The center of the 3D electron cloud

# Plot the single point in blue
fig = plt.figure(figsize=(10, 10))
ax = fig.add_subplot(111, projection='3d')

# Plot a single blue point
ax.scatter(X, Y, Z, color='blue', s=200)

# Adjust view for better visualization
ax.view_init(elev=30, azim=60)

plt.show()
```
