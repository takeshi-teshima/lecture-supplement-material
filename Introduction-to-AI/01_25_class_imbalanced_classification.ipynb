```python
#############################
# 1. Imports and Random Seeds
#############################
import random
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
import pytorch_lightning as pl
from pytorch_lightning.callbacks import EarlyStopping
from torch.utils.data import DataLoader, TensorDataset
from sklearn.datasets import make_blobs
from sklearn.model_selection import train_test_split
from sklearn.metrics import (
    confusion_matrix,
    precision_recall_fscore_support,
    accuracy_score
)
import matplotlib.pyplot as plt
from matplotlib.colors import ListedColormap

# Fix seeds
SEED = 94
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)
torch.backends.cudnn.deterministic = True
torch.backends.cudnn.benchmark = False

#############################
# 2. Color Palette
#############################
color_palette = {
    "white": "#FFFFFF",
    "light_gray": "#D3D3D3",
    "gray": "#808080",
    "black": "#000000",
    "green": "#008000",
    "blue": "#0000FF",
    "light_blue": "#ADD8E6",
    "light_light_blue": "#E0FFFF",
    "yellow": "#FFFF00",
    "orange": "#FFA500",
    "dark_orange": "#FF8C00",
    "purple": "#800080",
}

class_colors = [color_palette["blue"], color_palette["dark_orange"]]
region_cmap = ListedColormap(class_colors)

#############################
# 3. Generate Synthetic Data
#############################
# One large majority cluster + two small minority clusters
X_maj, _ = make_blobs(
    n_samples=2910,
    centers=[[0, 0]],
    cluster_std=[2.5],
    random_state=SEED
)
y_maj = np.zeros(2910, dtype=int)

X_min1, _ = make_blobs(
    n_samples=45,
    centers=[[1, -1]],
    cluster_std=[0.3],
    random_state=SEED + 1
)
y_min1 = np.ones(45, dtype=int)

X_min2, _ = make_blobs(
    n_samples=45,
    centers=[[-1, 1]],
    cluster_std=[0.3],
    random_state=SEED + 2
)
y_min2 = np.ones(45, dtype=int)

# Combine and shuffle
X = np.vstack([X_maj, X_min1, X_min2])
y = np.hstack([y_maj, y_min1, y_min2])
indices = np.arange(len(y))
np.random.shuffle(indices)
X = X[indices]
y = y[indices]

# Split: 70% train, 15% val, 15% test
X_temp, X_test, y_temp, y_test = train_test_split(X, y, test_size=0.15, random_state=SEED)
X_train, X_val, y_train, y_val = train_test_split(X_temp, y_temp, test_size=0.1765, random_state=SEED)

#############################
# 4. Visualization Functions
#############################
def plot_data_scatter(X_data, y_data, title="Data Scatter Plot"):
    plt.figure(figsize=(6,5))
    for cls, c in [(0, class_colors[0]), (1, class_colors[1])]:
        idxs = (y_data == cls)
        plt.scatter(X_data[idxs, 0], X_data[idxs, 1],
                    color=c, edgecolors=color_palette["black"],
                    label=f"Class {cls}")
    plt.title(title)
    plt.xlabel("Feature 1")
    plt.ylabel("Feature 2")
    plt.legend()
    plt.show()

def plot_decision_regions(model, X_data=None, y_data=None, title="Decision Boundary", show_data=True):
    """
    If X_data, y_data is None or show_data=False, it plots only the boundary.
    Otherwise, it overlays data points on top of the boundary.
    """
    if X_data is not None and len(X_data) > 0:
        x_min = X_data[:,0].min() - 2.0
        x_max = X_data[:,0].max() + 2.0
        y_min = X_data[:,1].min() - 2.0
        y_max = X_data[:,1].max() + 2.0
    else:
        x_min, x_max = -10, 10
        y_min, y_max = -10, 10

    step = 0.1
    xx, yy = np.meshgrid(np.arange(x_min, x_max, step),
                         np.arange(y_min, y_max, step))
    grid_points = np.c_[xx.ravel(), yy.ravel()]
    grid_tensor = torch.tensor(grid_points, dtype=torch.float32)

    model.eval()  # <--- calls eval() on either a single model or our ensemble wrapper
    with torch.no_grad():
        logits = model(grid_tensor)  # <--- calls model.forward(grid_tensor)
        probs = torch.sigmoid(logits).squeeze().cpu().numpy()
        preds = (probs >= 0.5).astype(int)
    Z = preds.reshape(xx.shape)

    plt.figure(figsize=(6,5))
    plt.contourf(xx, yy, Z, levels=[-0.5, 0.5, 1.5], cmap=region_cmap, alpha=0.3)
    if show_data and (X_data is not None) and (y_data is not None):
        for cls, c in [(0, class_colors[0]), (1, class_colors[1])]:
            idxs = (y_data == cls)
            plt.scatter(X_data[idxs, 0], X_data[idxs, 1],
                        color=c, edgecolors=color_palette["black"],
                        label=f"Class {cls}")
    plt.title(title)
    plt.xlabel("Feature 1")
    plt.ylabel("Feature 2")
    if show_data:
        plt.legend()
    plt.show()

#############################
# 5. DataModule (Lightning)
#############################
import pytorch_lightning as pl
from torch.utils.data import DataLoader, TensorDataset

X_train_t = torch.tensor(X_train, dtype=torch.float32)
y_train_t = torch.tensor(y_train, dtype=torch.float32).unsqueeze(1)
X_val_t   = torch.tensor(X_val,   dtype=torch.float32)
y_val_t   = torch.tensor(y_val,   dtype=torch.float32).unsqueeze(1)
X_test_t  = torch.tensor(X_test,  dtype=torch.float32)
y_test_t  = torch.tensor(y_test,  dtype=torch.float32).unsqueeze(1)

class BinaryDataModule(pl.LightningDataModule):
    def __init__(self, X_train, y_train, X_val, y_val, X_test, y_test, batch_size=32):
        super().__init__()
        self.batch_size = batch_size
        self.X_train = X_train
        self.y_train = y_train
        self.X_val   = X_val
        self.y_val   = y_val
        self.X_test  = X_test
        self.y_test  = y_test

    def setup(self, stage=None):
        self.train_dataset = TensorDataset(self.X_train, self.y_train)
        self.val_dataset   = TensorDataset(self.X_val, self.y_val)
        self.test_dataset  = TensorDataset(self.X_test, self.y_test)

    def train_dataloader(self):
        return DataLoader(self.train_dataset, batch_size=self.batch_size, shuffle=True)

    def val_dataloader(self):
        return DataLoader(self.val_dataset, batch_size=self.batch_size, shuffle=False)

    def test_dataloader(self):
        return DataLoader(self.test_dataset, batch_size=self.batch_size, shuffle=False)

#############################
# 6. Model Definition
#############################
class BinaryClassifier(pl.LightningModule):
    def __init__(self, input_dim=2, hidden1=64, hidden2=32, lr=1e-3):
        super().__init__()
        self.save_hyperparameters()
        self.model = nn.Sequential(
            nn.Linear(input_dim, hidden1),
            nn.ReLU(),
            nn.Linear(hidden1, hidden2),
            nn.ReLU(),
            nn.Linear(hidden2, 1)
        )
        self.criterion = nn.BCEWithLogitsLoss()

        # We'll store losses for each batch to compute epoch metrics manually
        self.train_losses = []
        self.val_losses = []

    def forward(self, x):
        return self.model(x)

    def training_step(self, batch, batch_idx):
        x, y = batch
        logits = self(x)
        loss = self.criterion(logits, y)
        self.train_losses.append(loss.item())
        return loss

    def validation_step(self, batch, batch_idx):
        x, y = batch
        logits = self(x)
        loss = self.criterion(logits, y)
        self.val_losses.append(loss.item())
        self.log("val_loss", loss, on_step=False, on_epoch=True)
        return loss

    def test_step(self, batch, batch_idx):
        x, y = batch
        logits = self(x)
        loss = self.criterion(logits, y)
        preds = (torch.sigmoid(logits) >= 0.5).float()
        acc = (preds == y).float().mean()
        return {"loss": loss, "acc": acc}

    def configure_optimizers(self):
        return optim.Adam(self.parameters(), lr=self.hparams.lr)

#############################
# Common Setup
#############################
max_epochs = 300
early_stopping_patience = 10
batch_size = 256

data_module = BinaryDataModule(X_train_t, y_train_t, X_val_t, y_val_t, X_test_t, y_test_t, batch_size=batch_size)

#############################
# 7. Train Baseline
#############################
early_stop_callback = EarlyStopping(monitor="val_loss", patience=early_stopping_patience, mode="min", verbose=False)
trainer_baseline = pl.Trainer(
    max_epochs=max_epochs,
    callbacks=[early_stop_callback],
    deterministic=True,
    logger=False,
    enable_checkpointing=False,
    enable_progress_bar=False
)

baseline_model = BinaryClassifier()
trainer_baseline.fit(baseline_model, datamodule=data_module)
baseline_test_results = trainer_baseline.test(baseline_model, datamodule=data_module, verbose=False)

#############################
# 8. Evaluate & Plot Baseline
#############################
def batch_losses_to_epoch_losses(batch_losses, num_batches_per_epoch):
    epoch_losses = []
    idx = 0
    while idx < len(batch_losses):
        epoch_losses.append(np.mean(batch_losses[idx : idx + num_batches_per_epoch]))
        idx += num_batches_per_epoch
    return epoch_losses

train_batches_per_epoch = len(data_module.train_dataloader())
val_batches_per_epoch   = len(data_module.val_dataloader())

baseline_train_loss_per_epoch = batch_losses_to_epoch_losses(baseline_model.train_losses, train_batches_per_epoch)
baseline_val_loss_per_epoch   = batch_losses_to_epoch_losses(baseline_model.val_losses,   val_batches_per_epoch)

def predict_dataset(model, X_data, y_data):
    model.eval()
    with torch.no_grad():
        X_t = torch.tensor(X_data, dtype=torch.float32)
        logits = model(X_t)
        probs = torch.sigmoid(logits).squeeze().numpy()
        preds = (probs >= 0.5).astype(int)
    return preds

from sklearn.metrics import (
    confusion_matrix,
    precision_recall_fscore_support,
    accuracy_score
)

def compute_metrics(y_true, y_pred):
    cm = confusion_matrix(y_true, y_pred)
    prec, rec, f1, _ = precision_recall_fscore_support(y_true, y_pred, average="binary", zero_division=0)
    acc = accuracy_score(y_true, y_pred)
    misclass = 1.0 - acc
    return {
        "confusion_matrix": cm,
        "precision": prec,
        "recall": rec,
        "f1": f1,
        "misclassification_rate": misclass
    }

# Evaluate on train/val/test
baseline_train_preds = predict_dataset(baseline_model, X_train, y_train)
baseline_val_preds   = predict_dataset(baseline_model, X_val,   y_val)
baseline_test_preds  = predict_dataset(baseline_model, X_test,  y_test)

baseline_train_metrics = compute_metrics(y_train, baseline_train_preds)
baseline_val_metrics   = compute_metrics(y_val,   baseline_val_preds)
baseline_test_metrics  = compute_metrics(y_test,  baseline_test_preds)

print("\n=== Baseline Model Metrics ===")
print("Train Loss (final epoch):", baseline_train_loss_per_epoch[-1])
print("Val   Loss (final epoch):", baseline_val_loss_per_epoch[-1])
print("Train Metrics:", baseline_train_metrics)
print("Val   Metrics:", baseline_val_metrics)
print("Test  Metrics:", baseline_test_metrics)

# Plot data
plot_data_scatter(X_train, y_train, title="Train Data Scatter Plot")

# Plot baseline boundary in separate figures
plot_decision_regions(baseline_model, None, None, title="Baseline (No Data)", show_data=False)
plot_decision_regions(baseline_model, X_train, y_train, title="Baseline (Train Data)", show_data=True)
plot_decision_regions(baseline_model, X_test,  y_test,  title="Baseline (Test Data)", show_data=True)

#############################
# 9. Underbagging
#############################
n_ensemble = 50
ensemble_models = []
sub_data_list = []

y_train_np = y_train
indices_minority = np.where(y_train_np == 1)[0]
indices_majority = np.where(y_train_np == 0)[0]
n_minority = len(indices_minority)

for i in range(n_ensemble):
    sampled_majority = np.random.choice(indices_majority, size=n_minority, replace=False)
    selected_indices = np.concatenate([indices_minority, sampled_majority])
    np.random.shuffle(selected_indices)

    X_sub = X_train[selected_indices]
    y_sub = y_train_np[selected_indices]
    sub_data_list.append((X_sub, y_sub))

    X_sub_t = torch.tensor(X_sub, dtype=torch.float32)
    y_sub_t = torch.tensor(y_sub, dtype=torch.float32).unsqueeze(1)
    sub_dataset = TensorDataset(X_sub_t, y_sub_t)
    sub_loader = DataLoader(sub_dataset, batch_size=batch_size, shuffle=True)

    sub_model = BinaryClassifier()
    trainer_sub = pl.Trainer(
        max_epochs=max_epochs,
        callbacks=[EarlyStopping(monitor="val_loss", patience=early_stopping_patience, mode="min", verbose=False)],
        deterministic=True,
        logger=False,
        enable_checkpointing=False,
        enable_progress_bar=False
    )
    trainer_sub.fit(sub_model, train_dataloaders=sub_loader, val_dataloaders=data_module.val_dataloader())
    ensemble_models.append(sub_model)

#############################
# 10. Ensemble Wrapper (Approach 2)
#############################
class EnsembleWrapper(nn.Module):
    """
    Wraps a list of BinaryClassifier sub-models into a single nn.Module
    that does forward pass by averaging their predicted probabilities.
    """
    def __init__(self, ensemble_models):
        super().__init__()
        # We store sub-models in a ModuleList so PyTorch recognizes them.
        # This is optional if we just want forward logic, but it's good practice.
        self.ensemble = nn.ModuleList(ensemble_models)

    def forward(self, x):
        # We'll compute the average predicted probability across sub-models,
        # then convert that to a "logit" so that the code in plot_decision_regions
        # can treat it as if it were a single model's logits.
        all_probs = []
        for m in self.ensemble:
            m.eval()  # ensure each sub-model is in eval mode
            logits = m(x)
            probs = torch.sigmoid(logits)
            all_probs.append(probs)

        # shape: (n_models, N, 1)
        all_probs_stack = torch.stack(all_probs, dim=0)
        avg_prob = torch.mean(all_probs_stack, dim=0)  # shape: (N,1)

        # Convert prob => logit = log(prob / (1-prob))
        eps = 1e-7
        avg_prob_clamped = torch.clamp(avg_prob, eps, 1.0 - eps)
        logits_ens = torch.log(avg_prob_clamped / (1.0 - avg_prob_clamped))
        return logits_ens

#############################
# 11. Evaluate Ensemble
#############################
ensemble_module = EnsembleWrapper(ensemble_models)

# We'll reuse predict_dataset(...) for train/val/test
def predict_ensemble_module(module, X_data):
    module.eval()
    with torch.no_grad():
        X_t = torch.tensor(X_data, dtype=torch.float32)
        logits = module(X_t)
        probs = torch.sigmoid(logits).squeeze().cpu().numpy()
        preds = (probs >= 0.5).astype(int)
    return preds

ensemble_train_preds = predict_ensemble_module(ensemble_module, X_train)
ensemble_val_preds   = predict_ensemble_module(ensemble_module, X_val)
ensemble_test_preds  = predict_ensemble_module(ensemble_module, X_test)

ensemble_train_metrics = compute_metrics(y_train, ensemble_train_preds)
ensemble_val_metrics   = compute_metrics(y_val,   ensemble_val_preds)
ensemble_test_metrics  = compute_metrics(y_test,  ensemble_test_preds)

print("\n=== Underbagging Ensemble Metrics ===")
print("Train Metrics:", ensemble_train_metrics)
print("Val   Metrics:", ensemble_val_metrics)
print("Test  Metrics:", ensemble_test_metrics)

#############################
# 12. Plot Ensemble Boundaries
#############################
plot_decision_regions(ensemble_module, None, None, title="Underbagging Ensemble (No Data)", show_data=False)
plot_decision_regions(ensemble_module, X_train, y_train, title="Underbagging Ensemble (Train Data)", show_data=True)
plot_decision_regions(ensemble_module, X_test,  y_test,  title="Underbagging Ensemble (Test Data)",  show_data=True)

#############################
# 13. Visualize Some Base Classifiers
#############################
def plot_some_base_classifiers(models, sub_data_list, num_submodels=3):
    """
    Randomly pick sub-models from the ensemble, and plot their
    decision boundary with the sub-dataset that trained them.
    """
    chosen_indices = np.random.choice(len(models), size=num_submodels, replace=False)
    for idx in chosen_indices:
        sub_model = models[idx]
        X_sub, y_sub = sub_data_list[idx]
        # Plot boundary with sub-dataset
        plot_decision_regions(
            sub_model,
            X_sub,
            y_sub,
            title=f"Base Classifier #{idx} on its Bootstrap Data"
        )

# Show 3 random base classifiers with their training subsets
plot_some_base_classifiers(ensemble_models, sub_data_list, num_submodels=3)
```
