```python
# Import necessary libraries
import matplotlib.pyplot as plt
from skimage import data, color, exposure
from skimage.feature import hog
import numpy as np

# Load a sample image from scikit-image
image = data.astronaut()

# Convert the image to grayscale
gray_image = color.rgb2gray(image)

# Compute HOG features and the corresponding HOG image
hog_features, hog_image = hog(
    gray_image,
    orientations=9,
    pixels_per_cell=(16, 16),
    cells_per_block=(2, 2),
    block_norm='L2-Hys',
    visualize=True,
    feature_vector=True
)

# Display the original image
plt.figure(figsize=(6, 6))
plt.axis('off')
plt.imshow(image)
plt.title('Original Image')
plt.show()

# Enhance the HOG image for better visualization
hog_image_rescaled = exposure.rescale_intensity(hog_image, in_range=(0, 10))

# Display the rescaled HOG visualization
plt.figure(figsize=(6, 6))
plt.axis('off')
plt.imshow(hog_image_rescaled, cmap='gray')
plt.title('HOG Visualization (Rescaled)')
plt.show()

# Print the first 10 elements of the HOG feature vector
print("First 10 elements of the HOG feature vector:")
print(np.round(hog_features, 2)[:10])
```
