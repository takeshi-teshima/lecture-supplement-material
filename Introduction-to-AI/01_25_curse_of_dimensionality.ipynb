```python
import numpy as np
import matplotlib.pyplot as plt

# -----------------------------------------
# Color Palette
# -----------------------------------------
color_palette = {
    "white": "#FFFFFF",
    "light_gray": "#D3D3D3",
    "gray": "#808080",
    "black": "#000000",
    "green": "#008000",
    "blue": "#0000FF",
    "light_blue": "#ADD8E6",
    "light_light_blue": "#E0FFFF",
    "yellow": "#FFFF00",
    "orange": "#FFA500",
    "dark_orange": "#FF8C00",
    "purple": "#800080",
}

# 固定乱数
np.random.seed(0)

# パラメータ
n_samples = 10000
batch_size = 1000
dimensions = [2, 1000]

for d in dimensions:
    print(f"Generating norms for {d}-dimensional normal distribution...")
    num_batches = n_samples // batch_size
    norms = []

    for _ in range(num_batches):
        X_batch = np.random.randn(batch_size, d)
        batch_norms = np.linalg.norm(X_batch, axis=1)
        norms.extend(batch_norms)

    norms = np.array(norms)

    # ヒストグラムの表示
    plt.figure(figsize=(8, 6), facecolor=color_palette["white"])
    plt.hist(norms, bins=100, color=color_palette["blue"], edgecolor=color_palette["blue"])
    plt.title(f"Distribution of norms in {d}-dimensional Normal distribution", fontsize=14)
    plt.xlabel("Norm", fontsize=12)
    plt.ylabel("Frequency", fontsize=12)
    plt.grid(True, linestyle='--', alpha=0.5)
    plt.xlim(left=-0.1)
    plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from matplotlib.colors import LinearSegmentedColormap

# Palette: Light Blue to Orange
blue = "#ADD8E6"
dark_orange = "#FF8C00"
orange = "#FFA500"
original_cmap = LinearSegmentedColormap.from_list(
    "blue_orange",
    [blue, dark_orange]
)

# Swiss roll parameters (opened structure)
t = np.linspace(1.5 * np.pi, 3.5 * np.pi, 60)
h = np.linspace(-10, 15, 30)
T, H = np.meshgrid(t, h)
X_surface = T * np.cos(T)
Y_surface = H
Z_surface = T * np.sin(T)

# Normalize for color mapping
norm_values = (T - T.min()) / (T.max() - T.min())

# 3D Plot
fig = plt.figure(figsize=(10, 6), facecolor='#FFFFFF')
ax = fig.add_subplot(111, projection='3d')

# Surface plot using lighting-style shading by reducing alpha for the back-facing areas
# Trick: simulate backside dimming by adjusting the shading and lighting
surface = ax.plot_surface(
    X_surface, Y_surface, Z_surface,
    rstride=1, cstride=1,
    facecolors=original_cmap(norm_values),
    linewidth=0,
    antialiased=True,
    shade=True,
    # shade=False,
    alpha=0.8  # uniform alpha, lighting will handle the back-side effect
)

# Light and view settings
# ax.set_title("Swiss Roll", fontsize=16, color="#000000")
ax.set_xlabel("X", color="#000000")
ax.set_ylabel("Y", color="#000000")
ax.set_zlabel("Z", color="#000000")
# ax.set_xticks([])
# ax.set_yticks([])
# ax.set_zticks([])
ax.view_init(elev=20, azim=120)

plt.tight_layout()
plt.show()
```
