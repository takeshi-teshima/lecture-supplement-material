```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_swiss_roll
from mpl_toolkits.mplot3d import Axes3D

# データ生成
n_samples = 2000
noise = 0.05
X, color = make_swiss_roll(n_samples=n_samples, noise=noise)

# パレットから色選択（Light Blue 〜 Purple にグラデーション）
from matplotlib.colors import ListedColormap
custom_cmap = ListedColormap(["#E0FFFF", "#ADD8E6", "#0000FF", "#800080"])

# 3Dプロット
fig = plt.figure(figsize=(10, 6), facecolor='#FFFFFF')
ax = fig.add_subplot(111, projection='3d')
sc = ax.scatter(X[:, 0], X[:, 1], X[:, 2], c=color, cmap=custom_cmap, s=10, alpha=0.8)

ax.set_title("Swiss Roll", fontsize=16, color="#000000")
ax.set_xlabel("X", color="#000000")
ax.set_ylabel("Y", color="#000000")
ax.set_zlabel("Z", color="#000000")

# 軸の色を調整
for spine in ax.spines.values():
    spine.set_color('#808080')

plt.show()
```
