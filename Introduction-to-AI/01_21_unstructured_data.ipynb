```python
# -*- coding: utf-8 -*-

# 入力文字列（例）
text = "いい天気だね"

# 各文字に対してUTF-8符号化を表示
for char in text:
    utf8_bytes = char.encode('utf-8')
    hex_repr = ' '.join(f'{byte:02x}' for byte in utf8_bytes)
    dec_repr = ' '.join(f'{byte:3d}' for byte in utf8_bytes)
    print(f"{char}: HEX: {hex_repr} | DEC: {dec_repr}")

print([f"{char}" for char in text])
print(text)
print([
    ''.join(f'{byte:3d}' for byte in char.encode('utf-8'))
    for char in text
])
```
