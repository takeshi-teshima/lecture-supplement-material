```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification, make_moons, make_blobs
from sklearn.neighbors import KNeighborsClassifier
from matplotlib.colors import ListedColormap
from scipy.spatial import Voronoi, voronoi_plot_2d

# 色設定
orange = "#FFA500"
blue = "#0000FF"
light_orange = "#FFEDD5"
light_blue = "#E0FFFF"
dark_orange = "#FF8C00"
black = "#000000"

background_cmap = ListedColormap([light_blue, light_orange])

def add_label_noise(y, noise_ratio=0.1, random_state=None):
    rng = np.random.RandomState(random_state)
    n_samples = len(y)
    n_noisy = int(n_samples * noise_ratio)
    indices = rng.choice(n_samples, size=n_noisy, replace=False)
    y_noisy = np.array(y, copy=True)
    y_noisy[indices] = 1 - y_noisy[indices]
    return y_noisy

def plot_dataset_only(X, y, xlim, ylim, title=""):
    plt.figure(figsize=(6, 5))
    plt.title(title)
    for class_value, marker, color in zip([0, 1], ['x', 'o'], [blue, orange]):
        plt.scatter(X[y == class_value, 0], X[y == class_value, 1],
                    c=color, marker=marker, edgecolor=black, s=70, label=f"Class {class_value}")
    plt.xlim(xlim)
    plt.ylim(ylim)
    plt.xlabel("Feature 1")
    plt.ylabel("Feature 2")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

def plot_single_knn(X, y, k, xlim, ylim, title_prefix="", show_voronoi=True):
    clf = KNeighborsClassifier(n_neighbors=k)
    clf.fit(X, y)

    h = 0.05
    xx, yy = np.meshgrid(np.arange(xlim[0], xlim[1], h),
                         np.arange(ylim[0], ylim[1], h))
    Z = clf.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

    plt.figure(figsize=(6, 5))
    # 背景（決定領域）
    plt.contourf(xx, yy, Z, cmap=background_cmap, alpha=0.6)
    # 境界線（decision boundary）
    plt.contour(xx, yy, Z, levels=[0.5], colors=[dark_orange], linewidths=3)

    # データ点
    for class_value, marker, color in zip([0, 1], ['x', 'o'], [blue, orange]):
        plt.scatter(X[y == class_value, 0], X[y == class_value, 1],
                    c=color, marker=marker, edgecolor=black, s=70)

    # Voronoi図（k=1のみ）
    if k == 1 and show_voronoi:
        vor = Voronoi(X)
        voronoi_plot_2d(vor, ax=plt.gca(), show_vertices=False,
                        line_colors=black, line_width=2, line_alpha=1.0, point_size=0)

    plt.title(f"k = {k}")
    # plt.title(f"{title_prefix}  k = {k}")
    plt.xlim(xlim)
    plt.ylim(ylim)
    plt.xlabel("Feature 1")
    plt.ylabel("Feature 2")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

def generate_and_plot_all(title_prefix, dataset_func, dataset_kwargs, label_noise=0.1):
    X, y_true = dataset_func(**dataset_kwargs)
    y = add_label_noise(y_true, noise_ratio=label_noise, random_state=42)

    # 共通のxlim, ylimを計算
    x_margin, y_margin = 1.0, 1.0
    x_min, x_max = X[:, 0].min() - x_margin, X[:, 0].max() + x_margin
    y_min, y_max = X[:, 1].min() - y_margin, X[:, 1].max() + y_margin
    xlim = (x_min, x_max)
    ylim = (y_min, y_max)

    plot_dataset_only(X, y, xlim, ylim, title=f"{title_prefix} Dataset")
    plot_single_knn(X, y, k=1, xlim=xlim, ylim=ylim, title_prefix=title_prefix, show_voronoi=False)
    plot_single_knn(X, y, k=1, xlim=xlim, ylim=ylim, title_prefix=title_prefix, show_voronoi=True)
    plot_single_knn(X, y, k=3, xlim=xlim, ylim=ylim, title_prefix=title_prefix)
    plot_single_knn(X, y, k=5, xlim=xlim, ylim=ylim, title_prefix=title_prefix)

# Blobs データセット
generate_and_plot_all(
    title_prefix="Blobs",
    dataset_func=make_blobs,
    dataset_kwargs=dict(
        n_samples=[20, 20],
        centers=[(-1, -1), (2, 2)],
        cluster_std=[0.8, 1.2],
        random_state=4
    ),
    label_noise=0.03
)

# Two Moons データセット
generate_and_plot_all(
    title_prefix="Two Moons",
    dataset_func=make_moons,
    dataset_kwargs=dict(
        n_samples=60,
        noise=0.25,
        random_state=10
    ),
    label_noise=0.03
)

# 線形分離データセット
generate_and_plot_all(
    title_prefix="Linearly Separable",
    dataset_func=make_classification,
    dataset_kwargs=dict(
        n_samples=30,
        n_features=2,
        n_redundant=0,
        n_clusters_per_class=1,
        n_classes=2,
        class_sep=1.5,
        flip_y=0.1,
        random_state=2
    ),
    label_noise=0
)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, zero_one_loss
from matplotlib.colors import ListedColormap

# Seed for reproducibility
np.random.seed(0)

# -----------------------------------------
# Original Color Palette
# -----------------------------------------
color_palette = {
    "white": "#FFFFFF",
    "light_gray": "#D3D3D3",
    "gray": "#808080",
    "black": "#000000",
    "green": "#008000",
    "blue": "#0000FF",
    "light_blue": "#ADD8E6",
    "light_light_blue": "#E0FFFF",
    "yellow": "#FFFF00",
    "orange": "#FFA500",
    "dark_orange": "#FF8C00",
    "purple": "#800080",
}

# Class Colors
class_colors = [
    color_palette["blue"],        # Class 0
    color_palette["dark_orange"], # Class 1
]

region_cmap = ListedColormap([
    color_palette["light_blue"], 
    color_palette["orange"]
])

# -----------------------------------------
# Generate Dataset
# -----------------------------------------
X_full, y_full = make_moons(n_samples=100, noise=0.2, random_state=42)
X_test, y_test = make_moons(n_samples=100, noise=0.2, random_state=24)

# Train-validation split without stratification
X_train, X_val, y_train, y_val = train_test_split(
    X_full, y_full, test_size=0.2, random_state=42
)

# Parameters
ks = np.arange(1, 10+1)
metrics = ['euclidean']
# metrics = ['euclidean', 'manhattan']
val_losses = {metric: [] for metric in metrics}

# -----------------------------------------
# Hyperparameter Selection
# -----------------------------------------
best_k, best_metric, best_acc = None, None, -np.inf
for metric in metrics:
    for k in ks:
        model = KNeighborsClassifier(n_neighbors=k, metric=metric)
        model.fit(X_train, y_train)
        y_val_pred = model.predict(X_val)
        acc = accuracy_score(y_val, y_val_pred)
        val_losses[metric].append(zero_one_loss(y_val, y_val_pred))

        if acc > best_acc:
            best_acc = acc
            best_k = k
            best_metric = metric

# -----------------------------------------
# Final Model
# -----------------------------------------
final_model = KNeighborsClassifier(n_neighbors=best_k, metric=best_metric)
final_model.fit(X_train, y_train)

# -----------------------------------------
# Visualization function
# -----------------------------------------
def plot_data(X, y, model=None, title="", show_decision_regions=False):
    plt.figure(figsize=(6, 5))
    
    if show_decision_regions and model is not None:
        x_min, x_max = X[:,0].min() - 0.5, X[:,0].max() + 0.5
        y_min, y_max = X[:,1].min() - 0.5, X[:,1].max() + 0.5
        xx, yy = np.meshgrid(np.linspace(x_min, x_max, 300),
                             np.linspace(y_min, y_max, 300))
        Z = model.predict(np.c_[xx.ravel(), yy.ravel()]).reshape(xx.shape)
        plt.contourf(xx, yy, Z, cmap=region_cmap, alpha=0.3)
        plt.contour(xx, yy, Z, levels=[0.5], colors=color_palette["orange"], linewidths=3)

    for cls in [0,1]:
        plt.scatter(X[y==cls,0], X[y==cls,1],
                    c=class_colors[cls],
                    marker='x' if cls==0 else 'o',
                    s=60, linewidth=1,
                    edgecolor=color_palette["black"],
                    label=f"Class {cls}")

    plt.title(title)
    plt.legend()
    plt.grid(True)
    plt.show()

# -----------------------------------------
# Plots without decision regions
# ----------------------------  -------------
plot_data(X_train, y_train, title="Training Data", show_decision_regions=False)
plot_data(X_test, y_test, title="Test Data", show_decision_regions=False)

# -----------------------------------------
# Decision regions for each metric and k (Train & Test)
# -----------------------------------------
for metric in metrics:
    for k in ks:
        model = KNeighborsClassifier(n_neighbors=k, metric=metric)
        model.fit(X_train, y_train)
        
        plot_data(X_train, y_train, model=model,
                  title=f"Train Decision Regions (k={k}, metric={metric})",
                  show_decision_regions=True)

        plot_data(X_test, y_test, model=model,
                  title=f"Test Decision Regions (k={k}, metric={metric})",
                  show_decision_regions=True)

# -----------------------------------------
# Best Decision Regions (train, valid, test)
# -----------------------------------------
plot_data(X_train, y_train, model=final_model,
          title=f"Training Data & Decision Region\n(Best k={best_k}, metric={best_metric})",
          show_decision_regions=True)

plot_data(X_val, y_val, model=final_model,
          title=f"Validation Data & Decision Region\n(Best k={best_k}, metric={best_metric})",
          show_decision_regions=True)

plot_data(X_test, y_test, model=final_model,
          title=f"Test Data & Decision Region\n(Best k={best_k}, metric={best_metric})",
          show_decision_regions=True)

# -----------------------------------------
# Validation loss plot (Separate figures, Common ylim)
# -----------------------------------------
ylim_max = max(max(losses) for losses in val_losses.values())

for metric in metrics:
    plt.figure(figsize=(8, 6))
    plt.plot(ks, val_losses[metric], marker='o', linewidth=3,
             color=color_palette["purple"])  # Euclideanもpurpleに統一
    plt.title(f"Validation Loss vs. k ({metric.capitalize()} metric)", fontsize=15)
    plt.xlabel("k", fontsize=13)
    plt.ylabel("Validation Loss (Misclassification Rate)", fontsize=13)
    plt.ylim(0, ylim_max + 0.1)
    plt.xticks(ks)
    plt.grid(True)
    plt.show()

# -----------------------------------------
# Final results
# -----------------------------------------
final_test_accuracy = accuracy_score(y_test, final_model.predict(X_test))
print(f"Best k: {best_k}, Best metric: {best_metric}, Validation Accuracy: {best_acc:.3f}")
print(f"Test Accuracy: {final_test_accuracy:.3f}")
```
