```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification, make_moons, make_blobs
from sklearn.neighbors import KNeighborsClassifier
from matplotlib.colors import ListedColormap
from scipy.spatial import Voronoi, voronoi_plot_2d

# 色設定
orange = "#FFA500"
blue = "#0000FF"
light_orange = "#FFEDD5"
light_blue = "#E0FFFF"
dark_orange = "#FF8C00"
black = "#000000"

background_cmap = ListedColormap([light_blue, light_orange])

def add_label_noise(y, noise_ratio=0.1, random_state=None):
    rng = np.random.RandomState(random_state)
    n_samples = len(y)
    n_noisy = int(n_samples * noise_ratio)
    indices = rng.choice(n_samples, size=n_noisy, replace=False)
    y_noisy = np.array(y, copy=True)
    y_noisy[indices] = 1 - y_noisy[indices]
    return y_noisy

def plot_dataset_only(X, y, xlim, ylim, title=""):
    plt.figure(figsize=(6, 5))
    plt.title(title)
    for class_value, marker, color in zip([0, 1], ['x', 'o'], [blue, orange]):
        plt.scatter(X[y == class_value, 0], X[y == class_value, 1],
                    c=color, marker=marker, edgecolor=black, s=70, label=f"Class {class_value}")
    plt.xlim(xlim)
    plt.ylim(ylim)
    plt.xlabel("Feature 1")
    plt.ylabel("Feature 2")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

def plot_single_knn(X, y, k, xlim, ylim, title_prefix="", show_voronoi=True):
    clf = KNeighborsClassifier(n_neighbors=k)
    clf.fit(X, y)

    h = 0.05
    xx, yy = np.meshgrid(np.arange(xlim[0], xlim[1], h),
                         np.arange(ylim[0], ylim[1], h))
    Z = clf.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

    plt.figure(figsize=(6, 5))
    # 背景（決定領域）
    plt.contourf(xx, yy, Z, cmap=background_cmap, alpha=0.6)
    # 境界線（decision boundary）
    plt.contour(xx, yy, Z, levels=[0.5], colors=[dark_orange], linewidths=3)

    # データ点
    for class_value, marker, color in zip([0, 1], ['x', 'o'], [blue, orange]):
        plt.scatter(X[y == class_value, 0], X[y == class_value, 1],
                    c=color, marker=marker, edgecolor=black, s=70)

    # Voronoi図（k=1のみ）
    if k == 1 and show_voronoi:
        vor = Voronoi(X)
        voronoi_plot_2d(vor, ax=plt.gca(), show_vertices=False,
                        line_colors=black, line_width=2, line_alpha=1.0, point_size=0)

    plt.title(f"k = {k}")
    # plt.title(f"{title_prefix}  k = {k}")
    plt.xlim(xlim)
    plt.ylim(ylim)
    plt.xlabel("Feature 1")
    plt.ylabel("Feature 2")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

def generate_and_plot_all(title_prefix, dataset_func, dataset_kwargs, label_noise=0.1):
    X, y_true = dataset_func(**dataset_kwargs)
    y = add_label_noise(y_true, noise_ratio=label_noise, random_state=42)

    # 共通のxlim, ylimを計算
    x_margin, y_margin = 1.0, 1.0
    x_min, x_max = X[:, 0].min() - x_margin, X[:, 0].max() + x_margin
    y_min, y_max = X[:, 1].min() - y_margin, X[:, 1].max() + y_margin
    xlim = (x_min, x_max)
    ylim = (y_min, y_max)

    plot_dataset_only(X, y, xlim, ylim, title=f"{title_prefix} - Dataset Only")
    plot_single_knn(X, y, k=1, xlim=xlim, ylim=ylim, title_prefix=title_prefix, show_voronoi=False)
    plot_single_knn(X, y, k=1, xlim=xlim, ylim=ylim, title_prefix=title_prefix, show_voronoi=True)
    plot_single_knn(X, y, k=3, xlim=xlim, ylim=ylim, title_prefix=title_prefix)
    plot_single_knn(X, y, k=5, xlim=xlim, ylim=ylim, title_prefix=title_prefix)

# Blobs データセット
generate_and_plot_all(
    title_prefix="Blobs",
    dataset_func=make_blobs,
    dataset_kwargs=dict(
        n_samples=[20, 20],
        centers=[(-1, -1), (2, 2)],
        cluster_std=[0.8, 1.2],
        random_state=4
    ),
    label_noise=0.03
)

# Two Moons データセット
generate_and_plot_all(
    title_prefix="Two Moons",
    dataset_func=make_moons,
    dataset_kwargs=dict(
        n_samples=60,
        noise=0.25,
        random_state=10
    ),
    label_noise=0.03
)

# 線形分離データセット
generate_and_plot_all(
    title_prefix="Linearly Separable",
    dataset_func=make_classification,
    dataset_kwargs=dict(
        n_samples=30,
        n_features=2,
        n_redundant=0,
        n_clusters_per_class=1,
        n_classes=2,
        class_sep=1.5,
        flip_y=0.1,
        random_state=2
    ),
    label_noise=0
)
```

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.neighbors import KNeighborsClassifier
from matplotlib.colors import ListedColormap

# Color setup
orange = "#FFA500"
blue = "#0000FF"
light_orange = "#FFEDD5"
light_blue = "#E0FFFF"
dark_orange = "#FF8C00"
black = "#000000"
background_cmap = ListedColormap([light_blue, light_orange])

def plot_knn_decision_boundary(X, y, k, xlim, ylim, title=""):
    clf = KNeighborsClassifier(n_neighbors=k)
    clf.fit(X, y)

    h = 0.05
    xx, yy = np.meshgrid(np.arange(xlim[0], xlim[1], h),
                         np.arange(ylim[0], ylim[1], h))
    Z = clf.predict(np.c_[xx.ravel(), yy.ravel()])
    Z = Z.reshape(xx.shape)

    plt.figure(figsize=(6, 5))
    plt.contourf(xx, yy, Z, cmap=background_cmap, alpha=0.6)
    plt.contour(xx, yy, Z, levels=[0.5], colors=[dark_orange], linewidths=3)

    for class_value, marker, color in zip([0, 1], ['x', 'o'], [blue, orange]):
        plt.scatter(X[y == class_value, 0], X[y == class_value, 1],
                    c=color, marker=marker, edgecolor=black, s=70)

    plt.title(title)
    plt.xlim(xlim)
    plt.ylim(ylim)
    # plt.xlabel("Feature 1")
    # plt.ylabel("Feature 2")
    plt.grid(True)
    plt.tight_layout()
    plt.show()

# Generate diagonal-aligned data
np.random.seed(42)
class0 = np.random.multivariate_normal([-2, -2], [[0.2, 0], [0, 0.2]], size=15)
class1 = np.random.multivariate_normal([2, 2], [[0.2, 0], [0, 0.2]], size=15)
X_orig = np.vstack([class0, class1])
y = np.array([0]*15 + [1]*15)

# Rescale feature 2 (vertical axis)
X_scaled = X_orig.copy()
X_scaled[:, 1] *= .1  # Large change

# Plot both versions
for label, X_plot in [("Original", X_orig), ("Rescaled", X_scaled)]:
    x_margin, y_margin = 1.0, 1.0
    x_min, x_max = X_plot[:, 0].min() - x_margin, X_plot[:, 0].max() + x_margin
    y_min, y_max = X_plot[:, 1].min() - y_margin, X_plot[:, 1].max() + y_margin
    xlim, ylim = (x_min, x_max), (y_min, y_max)
    ylim = -3.2, 3.2

    plot_knn_decision_boundary(X_plot, y, k=1, xlim=xlim, ylim=ylim,
                               title=f"{label} - kNN (k=1)")
    plot_knn_decision_boundary(X_plot, y, k=5, xlim=xlim, ylim=ylim,
                               title=f"{label} - kNN (k=5)")
```
